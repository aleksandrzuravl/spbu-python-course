# BIG MATRIX

```python
import numpy as np

n = 1500

x = np.linspace(0, 8*np.pi, n)
y = np.linspace(0, 8*np.pi, n)
X, Y = np.meshgrid(x, y)

matrix = (np.sin(X) * np.cos(3*Y) + 
          np.exp(-0.1*X) * np.sin(0.5*Y) + 
          0.3 * np.cos(2*X + Y) + 
          0.4 * np.tanh(0.2*X*Y) +
          0.1 * np.random.randn(n, n))

np.fill_diagonal(matrix, matrix.diagonal() + 2.0)


matrix[-1, -1] = 2.71828  

print(f"Размер матрицы: {matrix.shape}")
print(f"Количество элементов: {matrix.size:,}")
```
